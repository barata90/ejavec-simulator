```python
"""
EJAVEC 2026 — Validasi Prediction Interval & Out-of-Sample Test
Jalankan di cell SETELAH cell pengumpulan file.
Output disimpan ke FINAL_FOR_PAPER/tabel_body/
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import r2_score, mean_absolute_percentage_error
import os, warnings
warnings.filterwarnings('ignore')

ROOT = os.path.abspath('.')
NB11 = os.path.join(ROOT, 'nb11_fixed')
DST = os.path.join(ROOT, 'FINAL_FOR_PAPER')
os.makedirs(os.path.join(DST, 'tabel_body'), exist_ok=True)
os.makedirs(os.path.join(DST, 'gambar_body'), exist_ok=True)

print("=" * 70)
print("VALIDASI: Prediction Interval & Out-of-Sample Test")
print("=" * 70)

# ============================================================
# 1. LOAD DATA
# ============================================================
master = pd.read_csv(os.path.join(NB11, 'merged_yearly.csv'))
print(f"\n[LOAD] Master: {master.shape}, Kab: {master.Kabupaten_Kota.nunique()}")

df = master.copy()

# ============================================================
# 2. SETUP TARGETS & FEATURES
# ============================================================
TARGETS = {
    'Produksi_Ton': 'Produksi Padi (Ton)',
    'PDRB_Pertanian_MiliarRp': 'PDRB Pertanian (Miliar Rp)',
}

exclude = [
    'Kabupaten_Kota', 'Tahun', 'Region_Cluster', 'Prod_PerCapita',
    'PDRB_PerKapita_JutaRp', 'Share_Pertanian_Pct', 'Share_Industri_Pct',
    'Share_Perdagangan_Pct', 'FSI', 'Hilirisasi_Ratio', 'Yield_Ton_Ha',
    'TPT_Persen', 'Prod_GKG_Ton',
    'PDRB_Total_MiliarRp_growth', 'PDRB_Pertanian_MiliarRp_growth',
    'Produksi_Ton_growth',
] + list(TARGETS.keys())

# ============================================================
# 3. PER TARGET: OOS + ROLLING CV + PREDICTION INTERVAL
# ============================================================
all_oos = {}
all_rolling = {}
all_intervals = {}

for target_col, target_name in TARGETS.items():
    print(f"\n{'='*60}")
    print(f"  TARGET: {target_name}")
    print(f"{'='*60}")

    features = [c for c in df.columns
                if c not in exclude
                and df[c].dtype in ['float64', 'int64']
                and c != target_col]

    sub = df[['Kabupaten_Kota', 'Tahun', target_col] + features].dropna(subset=[target_col])
    for f in features:
        sub[f] = sub[f].fillna(sub[f].median())
    valid_f = [f for f in features if f in sub.columns and sub[f].notna().sum() > 10]

    # ---- A. OUT-OF-SAMPLE 2024-2025 ----
    print("\n  [A] Out-of-Sample Test (Train ≤2023, Test 2024-2025)")
    train = sub[sub['Tahun'] <= 2023]
    test = sub[sub['Tahun'] >= 2024]

    model = GradientBoostingRegressor(
        n_estimators=300, max_depth=5, learning_rate=0.05,
        subsample=0.8, random_state=42)
    model.fit(train[valid_f].values, train[target_col].values)

    y_pred_tr = model.predict(train[valid_f].values)
    y_pred_te = model.predict(test[valid_f].values)
    y_test = test[target_col].values

    r2_tr = r2_score(train[target_col].values, y_pred_tr)
    r2_te = r2_score(y_test, y_pred_te)
    mape_te = mean_absolute_percentage_error(y_test, y_pred_te) * 100
    rmse_te = np.sqrt(np.mean((y_test - y_pred_te) ** 2))

    print(f"      Train R² = {r2_tr:.4f}")
    print(f"      Test  R² = {r2_te:.4f}  MAPE = {mape_te:.1f}%  RMSE = {rmse_te:,.0f}")
    print(f"      Train N  = {len(train)} ({train.Tahun.min()}-{train.Tahun.max()})")
    print(f"      Test  N  = {len(test)} ({test.Tahun.min()}-{test.Tahun.max()})")

    # Per-kabupaten OOS error
    test_detail = test[['Kabupaten_Kota', 'Tahun', target_col]].copy()
    test_detail['Prediksi'] = y_pred_te
    test_detail['Error_Pct'] = np.abs(test_detail[target_col] - test_detail['Prediksi']) / (test_detail[target_col] + 1e-9) * 100

    all_oos[target_col] = {
        'r2_train': r2_tr, 'r2_test': r2_te,
        'mape': mape_te, 'rmse': rmse_te,
        'detail': test_detail, 'model': model,
        'features': valid_f,
    }

    # ---- B. ROLLING WINDOW CV (7 splits) ----
    print("\n  [B] Rolling Window Cross-Validation")
    years = sorted(sub['Tahun'].unique())
    cv_results = []

    for split_yr in range(2018, max(years)):
        tr = sub[sub['Tahun'] <= split_yr]
        te = sub[sub['Tahun'] == split_yr + 1]
        if len(te) < 5:
            continue

        m = GradientBoostingRegressor(
            n_estimators=200, max_depth=5, learning_rate=0.05,
            subsample=0.8, random_state=42)
        m.fit(tr[valid_f].values, tr[target_col].values)
        y_p = m.predict(te[valid_f].values)
        y_a = te[target_col].values

        r2 = r2_score(y_a, y_p)
        mape = mean_absolute_percentage_error(y_a, y_p) * 100
        rmse = np.sqrt(np.mean((y_a - y_p) ** 2))

        # Per-kab errors for interval estimation
        pct_errors = np.abs(y_a - y_p) / (np.abs(y_a) + 1e-9) * 100

        cv_results.append({
            'Split_Year': split_yr,
            'Test_Year': split_yr + 1,
            'R2': round(r2, 4),
            'MAPE': round(mape, 2),
            'RMSE': round(rmse, 1),
            'N_train': len(tr),
            'N_test': len(te),
            'MAPE_Std_PerKab': round(np.std(pct_errors), 2),
        })
        print(f"      Train ≤{split_yr} → Test {split_yr+1}: "
              f"R²={r2:.4f}  MAPE={mape:.1f}%  N_test={len(te)}")

    cv_df = pd.DataFrame(cv_results)
    mean_r2 = cv_df['R2'].mean()
    std_r2 = cv_df['R2'].std()
    mean_mape = cv_df['MAPE'].mean()
    std_mape = cv_df['MAPE'].std()

    print(f"\n      SUMMARY: R² = {mean_r2:.4f} ± {std_r2:.4f}")
    print(f"               MAPE = {mean_mape:.1f}% ± {std_mape:.1f}%")

    all_rolling[target_col] = cv_df

    # ---- C. PREDICTION INTERVAL ----
    print("\n  [C] Prediction Interval Estimation")

    # Metode: dari distribusi MAPE per rolling split
    mape_values = cv_df['MAPE'].values
    mape_mean = mape_values.mean()
    mape_std = mape_values.std()

    # 90% CI: mean ± 1.645 * std
    ci_lower_mape = mape_mean - 1.645 * mape_std
    ci_upper_mape = mape_mean + 1.645 * mape_std

    # Translate to growth interval
    # Baseline growth = 5.23%
    baseline_growth = 5.23
    growth_uncertainty = 1.645 * mape_std / 100 * baseline_growth  # proportional

    # Simpler: use MAPE std as direct uncertainty on growth (in pp)
    # More robust: from actual prediction errors on PDRB
    if target_col == 'PDRB_Pertanian_MiliarRp':
        # Collect actual growth prediction errors
        growth_errors = []
        for split_yr in range(2018, max(years)):
            tr = sub[sub['Tahun'] <= split_yr]
            te = sub[sub['Tahun'] == split_yr + 1]
            prev = sub[sub['Tahun'] == split_yr]
            if len(te) < 5 or len(prev) < 5:
                continue
            m = GradientBoostingRegressor(
                n_estimators=200, max_depth=5, learning_rate=0.05,
                subsample=0.8, random_state=42)
            m.fit(tr[valid_f].values, tr[target_col].values)

            for kab in te['Kabupaten_Kota'].unique():
                te_k = te[te['Kabupaten_Kota'] == kab]
                prev_k = prev[prev['Kabupaten_Kota'] == kab]
                if len(te_k) == 0 or len(prev_k) == 0:
                    continue
                actual_val = te_k[target_col].values[0]
                pred_val = m.predict(te_k[valid_f].values)[0]
                prev_val = prev_k[target_col].values[0]
                if prev_val > 0:
                    actual_growth = (actual_val / prev_val - 1) * 100
                    pred_growth = (pred_val / prev_val - 1) * 100
                    growth_errors.append(pred_growth - actual_growth)

        if growth_errors:
            ge = np.array(growth_errors)
            growth_std = np.std(ge)
            growth_bias = np.mean(ge)
            ci90_low = baseline_growth - 1.645 * growth_std
            ci90_high = baseline_growth + 1.645 * growth_std
            print(f"      Growth prediction error: mean={growth_bias:+.2f}pp, std={growth_std:.2f}pp")
            print(f"      Baseline growth: {baseline_growth:.2f}%")
            print(f"      90% Prediction Interval: [{ci90_low:.1f}%, {ci90_high:.1f}%]")

            all_intervals[target_col] = {
                'baseline': baseline_growth,
                'std_pp': round(growth_std, 2),
                'bias_pp': round(growth_bias, 2),
                'ci90_low': round(ci90_low, 1),
                'ci90_high': round(ci90_high, 1),
                'n_errors': len(growth_errors),
            }

    all_intervals[target_col + '_mape'] = {
        'mean': round(mape_mean, 1),
        'std': round(mape_std, 1),
        'ci90_low': round(ci_lower_mape, 1),
        'ci90_high': round(ci_upper_mape, 1),
    }

# ============================================================
# 4. SAVE OUTPUTS
# ============================================================
print(f"\n{'='*70}")
print("SAVING OUTPUTS")
print(f"{'='*70}")

# 4a. Rolling CV tables
for target_col, cv_df in all_rolling.items():
    fname = f'rolling_cv_{target_col}.csv'
    cv_df.to_csv(os.path.join(DST, 'tabel_body', fname), index=False)
    print(f"  ✅ {fname}")

# 4b. OOS summary table
oos_rows = []
for target_col, data in all_oos.items():
    oos_rows.append({
        'Target': TARGETS[target_col],
        'Train_Period': '2010-2023',
        'Test_Period': '2024-2025',
        'R2_Train': round(data['r2_train'], 4),
        'R2_Test_OOS': round(data['r2_test'], 4),
        'MAPE_Test': round(data['mape'], 1),
        'RMSE_Test': round(data['rmse'], 1),
    })
oos_df = pd.DataFrame(oos_rows)
oos_df.to_csv(os.path.join(DST, 'tabel_body', 'oos_validation_2024_2025.csv'), index=False)
print(f"  ✅ oos_validation_2024_2025.csv")

# 4c. Prediction interval summary
pi_rows = []
for key, val in all_intervals.items():
    pi_rows.append({'Metric': key, **val})
pi_df = pd.DataFrame(pi_rows)
pi_df.to_csv(os.path.join(DST, 'tabel_body', 'prediction_interval.csv'), index=False)
print(f"  ✅ prediction_interval.csv")

# 4d. Per-kab OOS detail (for appendix)
for target_col, data in all_oos.items():
    fname = f'oos_detail_{target_col}.csv'
    data['detail'].to_csv(os.path.join(DST, 'tabel_body', fname), index=False)
    print(f"  ✅ {fname}")

# ============================================================
# 5. FIGURES
# ============================================================
print(f"\n{'='*70}")
print("GENERATING FIGURES")
print(f"{'='*70}")

PAL = ['#2ecc71', '#3498db', '#e74c3c', '#f39c12', '#9b59b6']

# ---- 5a. OOS Scatter: Aktual vs Prediksi (2024-2025) ----
fig, axes = plt.subplots(1, 2, figsize=(13, 5.5))
for idx, (target_col, data) in enumerate(all_oos.items()):
    ax = axes[idx]
    det = data['detail']
    y_actual = det[target_col].values
    y_pred = det['Prediksi'].values

    ax.scatter(y_actual, y_pred, alpha=0.6, s=40, color=PAL[idx], edgecolors='white', linewidth=0.5)

    # Perfect fit line
    lims = [min(y_actual.min(), y_pred.min()) * 0.9,
            max(y_actual.max(), y_pred.max()) * 1.1]
    ax.plot(lims, lims, 'k--', alpha=0.5, linewidth=1, label='Perfect fit')

    # Regression line
    z = np.polyfit(y_actual, y_pred, 1)
    ax.plot(lims, np.polyval(z, lims), color=PAL[idx], alpha=0.8, linewidth=2)

    ax.set_xlabel(f'Aktual ({TARGETS[target_col].split("(")[0].strip()})')
    ax.set_ylabel('Prediksi')
    ax.set_title(f'{TARGETS[target_col]}\nR²={data["r2_test"]:.4f}, MAPE={data["mape"]:.1f}%',
                 fontweight='bold', fontsize=11)
    ax.legend(fontsize=8)
    ax.set_xlim(lims)
    ax.set_ylim(lims)

fig.suptitle('Validasi Out-of-Sample (2024–2025): Aktual vs Prediksi GBR',
             fontsize=13, fontweight='bold', y=1.02)
plt.tight_layout()
fig.savefig(os.path.join(DST, 'gambar_body', 'fig_oos_validation.png'),
            dpi=300, bbox_inches='tight', facecolor='white')
plt.close(fig)
print("  ✅ fig_oos_validation.png")

# ---- 5b. Rolling CV: R² & MAPE across splits ----
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# Panel a: R² over time
ax = axes[0]
for idx, (target_col, cv_df) in enumerate(all_rolling.items()):
    short = TARGETS[target_col].split('(')[0].strip()
    ax.plot(cv_df['Test_Year'], cv_df['R2'], 'o-', color=PAL[idx],
            linewidth=2, markersize=7, label=short)
    # Shaded band ± 1 std
    mean_r2 = cv_df['R2'].mean()
    std_r2 = cv_df['R2'].std()
    ax.axhline(mean_r2, color=PAL[idx], linestyle=':', alpha=0.4)
    ax.fill_between(cv_df['Test_Year'],
                    mean_r2 - std_r2, mean_r2 + std_r2,
                    alpha=0.1, color=PAL[idx])
ax.set_xlabel('Tahun Test')
ax.set_ylabel('R²')
ax.set_title('(a) Stabilitas R² Sepanjang Waktu', fontweight='bold')
ax.legend(fontsize=9)
ax.set_ylim(0.97, 1.005)
ax.axvspan(2020.5, 2021.5, alpha=0.1, color='red', label='COVID-19')

# Panel b: MAPE over time
ax = axes[1]
for idx, (target_col, cv_df) in enumerate(all_rolling.items()):
    short = TARGETS[target_col].split('(')[0].strip()
    ax.plot(cv_df['Test_Year'], cv_df['MAPE'], 's-', color=PAL[idx],
            linewidth=2, markersize=7, label=short)
    mean_mape = cv_df['MAPE'].mean()
    ax.axhline(mean_mape, color=PAL[idx], linestyle=':', alpha=0.4)
ax.set_xlabel('Tahun Test')
ax.set_ylabel('MAPE (%)')
ax.set_title('(b) Stabilitas MAPE Sepanjang Waktu', fontweight='bold')
ax.legend(fontsize=9)
ax.axvspan(2020.5, 2021.5, alpha=0.1, color='red')

fig.suptitle('Rolling Window Cross-Validation: 7 Split Temporal (2019–2025)',
             fontsize=13, fontweight='bold', y=1.02)
plt.tight_layout()
fig.savefig(os.path.join(DST, 'gambar_body', 'fig_rolling_cv.png'),
            dpi=300, bbox_inches='tight', facecolor='white')
plt.close(fig)
print("  ✅ fig_rolling_cv.png")

# ---- 5c. Prediction Interval Fan Chart ----
if 'PDRB_Pertanian_MiliarRp' in all_intervals:
    pi = all_intervals['PDRB_Pertanian_MiliarRp']

    fig, ax = plt.subplots(figsize=(10, 5.5))

    # Historical growth
    jatim = master.groupby('Tahun')['PDRB_Total_MiliarRp'].sum().reset_index()
    jatim['Growth'] = jatim['PDRB_Total_MiliarRp'].pct_change() * 100
    hist = jatim[jatim['Tahun'] >= 2016].dropna()

    ax.plot(hist['Tahun'], hist['Growth'], 'o-', color='#2c3e50',
            linewidth=2.5, markersize=8, label='Aktual', zorder=5)

    # 2026 projection with fan
    last_yr = hist['Tahun'].max()
    baseline = 5.23
    std = pi['std_pp']

    proj_x = [last_yr, last_yr + 1]
    proj_y = [hist[hist['Tahun'] == last_yr]['Growth'].values[0], baseline]
    ax.plot(proj_x, proj_y, 'o--', color='#e74c3c', linewidth=2, markersize=8,
            label=f'Proyeksi Baseline ({baseline:.2f}%)', zorder=5)

    # Fan bands
    bands = [
        (1.645, '90%', 0.10),  # 90% CI
        (1.282, '80%', 0.15),
        (1.000, '68%', 0.20),
        (0.675, '50%', 0.25),
    ]
    for z_val, label, alpha in bands:
        low = baseline - z_val * std
        high = baseline + z_val * std
        ax.fill_between([last_yr + 0.8, last_yr + 1.2],
                        [low, low], [high, high],
                        alpha=alpha, color='#e74c3c', zorder=2)
        if z_val == 1.645:
            ax.annotate(f'PI 90%: [{low:.1f}%, {high:.1f}%]',
                        xy=(last_yr + 1.0, high + 0.15),
                        fontsize=9, ha='center', color='#c0392b', fontweight='bold')

    # Scenario markers
    scenarios = [
        (2.13, 'Pesimistis\n(Perang)', '#e74c3c', 'v'),
        (6.03, 'Optimistis', '#27ae60', '^'),
        (8.00, 'Astacita', '#f39c12', 'D'),
    ]
    for g, lbl, clr, mkr in scenarios:
        ax.scatter(last_yr + 1, g, marker=mkr, s=100, color=clr, zorder=6, edgecolors='white')
        ax.annotate(lbl, xy=(last_yr + 1.15, g), fontsize=8, color=clr, va='center')

    ax.set_xlabel('Tahun', fontsize=11)
    ax.set_ylabel('Pertumbuhan PDRB (%)', fontsize=11)
    ax.set_title('Proyeksi Pertumbuhan PDRB Jawa Timur 2026\ndengan Prediction Interval dari Rolling CV',
                 fontsize=12, fontweight='bold')
    ax.legend(loc='upper left', fontsize=9)
    ax.set_xlim(2015.5, last_yr + 1.8)
    ax.set_ylim(-3, 10)
    ax.axhline(0, color='gray', linestyle='-', alpha=0.3)
    ax.grid(True, alpha=0.3)

    plt.tight_layout()
    fig.savefig(os.path.join(DST, 'gambar_body', 'fig_prediction_interval.png'),
                dpi=300, bbox_inches='tight', facecolor='white')
    plt.close(fig)
    print("  ✅ fig_prediction_interval.png")

# ============================================================
# 6. RINGKASAN UNTUK PAPER
# ============================================================
print(f"\n{'='*70}")
print("RINGKASAN ANGKA UNTUK PAPER")
print(f"{'='*70}")

for target_col, name in TARGETS.items():
    data = all_oos[target_col]
    cv = all_rolling[target_col]
    print(f"\n  {name}:")
    print(f"    OOS 2024-2025: R²={data['r2_test']:.4f}, MAPE={data['mape']:.1f}%")
    print(f"    Rolling CV:    R²={cv['R2'].mean():.4f} ± {cv['R2'].std():.4f}")
    print(f"                   MAPE={cv['MAPE'].mean():.1f}% ± {cv['MAPE'].std():.1f}%")

if 'PDRB_Pertanian_MiliarRp' in all_intervals:
    pi = all_intervals['PDRB_Pertanian_MiliarRp']
    print(f"\n  Prediction Interval (Pertumbuhan PDRB 2026):")
    print(f"    Baseline: {pi['baseline']:.2f}%")
    print(f"    Std error: ±{pi['std_pp']:.2f} pp")
    print(f"    90% CI: [{pi['ci90_low']:.1f}%, {pi['ci90_high']:.1f}%]")
    print(f"    Bias: {pi['bias_pp']:+.2f} pp (dari {pi['n_errors']} observasi)")

print(f"\n{'='*70}")
print("OUTPUT FILES:")
print(f"{'='*70}")
for subdir in ['tabel_body', 'gambar_body']:
    path = os.path.join(DST, subdir)
    for f in sorted(os.listdir(path)):
        size = os.path.getsize(os.path.join(path, f)) / 1024
        print(f"  {subdir}/{f:45s} {size:>8.0f} KB")
print(f"\n✅ SELESAI — Semua output prediction interval siap untuk paper")
```
